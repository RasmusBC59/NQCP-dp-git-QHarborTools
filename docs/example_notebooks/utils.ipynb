# Example of how to use the Utilitie functions for QHarbor

```python
import shutil
import sys
import tempfile
from pathlib import Path

from qharbortools.utils import (
    export_conda_environment,
    get_githash_of_current_commit,
    get_name_of_current_repo,
    git_info_to_file,
)
```

## Exporting the current Conda environment
When conducting experiments or analyses, it's often helpful to save the Python environment used to run your code. This ensures that you can reproduce your results later. If you're using Conda as your package manager, you can use the .export_conda_environment function to export the current environment to a file.


```python
# Create a temporary directory that persists
temp_dir = tempfile.mkdtemp()
temp_dir_path = Path(temp_dir)
print(f"Temporary directory created: {temp_dir_path}")


export_conda_environment(temp_dir_path)
```

For guidance on how to recreate an environment from an exported file, refer to the [Conda documentation](https://docs.conda.io/projects/conda/en/latest/user-guide/tasks/manage-environments.html#exporting-an-environment).

To demonstrate what an environment file looks like, we read it in and print its contents.


```python
from ruamel.yaml import YAML

yaml_file = temp_dir_path / "environment.yml"
with open(yaml_file) as file:
    yaml = YAML(typ="rt")

    yaml_content = yaml.load(file)
yaml.dump(yaml_content, sys.stdout)
```

## Saving the git hash and repository name
When working on a project, it's often useful to keep track of the specific version of the codebase you're using. This can be done by saving the git hash and repository name. The git_info_to_file function allows you to do this easily.

```python
git_info_to_file(temp_dir_path)
```

We can also read the saved git information from the file and print it to verify that it has been saved correctly.

```python
git_info_file = temp_dir_path / "git_info.txt"
with open(git_info_file) as file:
    git_info_content = file.read()
print(git_info_content)
```

We can also get the git hash and repository name directly, in the case we whant to add the information to another file or use it in a different context.

```python
print(f"Git Hash: {get_githash_of_current_commit()}")
print(f"Name of current repo: {get_name_of_current_repo()}")
```

```python
# Remember to clean up later when you're done:
shutil.rmtree(temp_dir_path)
```
