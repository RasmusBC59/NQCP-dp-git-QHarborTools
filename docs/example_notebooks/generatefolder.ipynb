# Generate folders for QHarbor

## 1. Creating a Conda Environment for Python 3.12

1. **Create a New Environment**

`conda create -n toolsqharbor python=3.12 -y`

2. **Change to newly made environment and activate it with Conda**

`conda activate toolsqharbor`

3. **Install Required Packages into new created enviroment**  
    To have access to DevAzure released packages, refer to [Installing NQCP Packages - Overview](https://dev.azure.com/NQCP/NQCP/_wiki/wikis/NQCP.wiki/375/Installing-NQCP-Packages).
    You can install packages that are needed:

`pip install qharhbortools`

---

## 2. Using generatefolder

1. **Start the app**  
    Double click on `generatefolders.bat` that is located in N:\SCI-NBI-NQCP\DataTeam.
    You can also make a copy of `generatefolders.bat` and place it where it is convinient, e.g the deskop.
    The interface for the Generate Folders tool should pop up:

<br>

![Generate Folders GUI](../_static/generatefolders_gui.png)

<br>

2. Click **Select Root Folder** and chose the folder where the folders and subfolders should be generated. 

3. Click **Select Info Folder** and pick excel file for data generation. The excel file should have the same structure as folderinfo.xlsx found in from N:\SCI-NBI-NQCP\DataTeam.

4. Click **Generate Folder** this will generate folders with datasets named '_QH_dataset_info.yaml'. 

Note:   *loading can take up to couple minutes , when double clicking on generatefolders.bat*<br>
    ![Generate Folders GUI](../_static/folder_Ndrive.png)
